# Brent Crude Oil Bollinger Bands


This notebook builds Bollinger Bands for Brent crude oil futures (`BZ=F`) and visualizes how closing prices move relative to a rolling mean plus or minus three standard deviations.


```python
import pandas as pd
import yfinance

from matplotlib import pyplot as plt

%matplotlib inline
```

```python
def sigma_bands(data, window, num_std=3):
    result = data.copy()
    moving_average = result["Close"].rolling(window=window).mean()
    rolling_std = result["Close"].rolling(window=window).std()
    result["UpperBB"] = moving_average + (num_std * rolling_std)
    result["LowerBB"] = moving_average - (num_std * rolling_std)
    return result

window = 50
asset_name = "Brent"
raw_data = yfinance.download("BZ=F", start="1962-01-01", end="2021-09-17", multi_level_index=False)

data = pd.DataFrame(raw_data)
bollinger_data = sigma_bands(data, window)
plot_data = pd.concat(
    [bollinger_data["Close"], bollinger_data["UpperBB"], bollinger_data["LowerBB"]],
    axis=1,
)

fig, ax = plt.subplots(figsize=(17, 6))

ax.plot(plot_data["Close"], linewidth=0.7, color="red", label=f"{asset_name} close price")
ax.plot(plot_data["UpperBB"], linewidth=0.7, color="blue", label="upper band")
ax.plot(plot_data["LowerBB"], linewidth=0.7, color="green", label="lower band")

ax.set_xlim(
    left=min(plot_data.index).to_pydatetime().date(),
    right=max(plot_data.index).to_pydatetime().date(),
)
ax.set_xlabel("Year")
ax.set_ylabel(f"{asset_name} price")
ax.set_title(f"{asset_name} Bollinger Bands ({window}-day window)")
ax.legend(loc="upper left")
ax.grid()
plt.show()
```
