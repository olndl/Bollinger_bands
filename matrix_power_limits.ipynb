# Matrix Power Limits

This notebook implements a small numerical routine that repeatedly raises a square matrix to higher powers and returns the first iteration that satisfies one of two threshold conditions.


## Implementation and Tests

The function validates that the matrix is square, tracks powers up to 1,000 iterations, and includes lightweight assertions for growth, decay, non-convergence, and invalid input cases.


```python
import numpy as np

def matrix_limes(A):
    """
    Iteratively calculating the power of a square matrix A until one of the conditions is met

    Conditions :
    1. For all pairs (i, j): A^n[i, j] < 0.01 * A^0[j, i] => return n
    2. There exists a pair (i, j): A^n[j, i] > 100 * A^0[j, i] => return -n
    If neither condition is satisfied after 1000 iteration, return false
    """
       
    A = np.array(A, dtype=float)
    max_iter=1000
    
    if A.shape[0] != A.shape[1]:
        raise ValueError("The matrix must be square")

    A_0 = A.copy()
    A_n = A.copy()
    A_0_T = A.T.copy() 
    
    for n in range(1, max_iter + 1):
        # Checking condition 1:
        if np.all(A_n < 0.01 * A_0_T):
            return n
        # Checking condition 2:
        if np.any(A_n > 100 * A_0):
            return -n
        A_n = np.dot(A_n, A)
    
    return False
```

```python
def run_tests():
    # Test 1
    A1 = np.array([
        [1, 2],
        [3, 4]
    ])
    result1 = matrix_limes(A1)
    assert result1 == -4, f"Test 1 failed: expected -4, got {result1}"
    print("Test 1 passed")

    # Test 2
    A2 = np.array([
        [0.5, 0.1],
        [0.2, 0.4]
    ])
    result2 = matrix_limes(A2)
    assert result2 == 13, f"Test 2 failed: expected 13, got {result2}"
    print("Test 2 passed")

    # Test 3
    A3 = np.array([
        [1, 0],
        [0, 1]
    ])
    result3 = matrix_limes(A3)
    assert result3 is False, f"Test 3 failed: expected False, got {result3}"
    print("Test 3 passed")

    # Test 4
    A_invalid = np.array([
        [1, 2, 3],
        [4, 5, 6]
    ])

    try:
        matrix_limes(A_invalid)
        assert False, "Test 4 failed: ValueError was not raised"
    except ValueError as e:
        assert str(e) == "The matrix must be square", \
            f"Test 4 failed: unexpected error message '{e}'"
        print("Test 4 passed")

   

    print("\nAll tests passed")



run_tests()
```
